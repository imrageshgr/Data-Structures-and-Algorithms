Create a list of 10 student marks.

Your program should:
- Find the highest mark.
- Find the lowest mark.
- Calculate the average.
- Count how many students passed (marks greater than or equal to 35).

```python
ten_students_marks = [85, 90, 75, 30, 75, 88, 66, 78, 34, 91]

print("highest: " ,max(ten_students_marks))

print("lowest: " ,min(ten_students_marks))

print("average: ", sum(ten_students_marks) / len(ten_students_marks) )

ten_students_marks.sort()
print(ten_students_marks)

count = 0 
for mark in ten_students_marks:
    if mark >= 35:
        count += 1

print("Number of students who scored greater than or equal to 35:", count)
```

Create a new list containing only even numbers.



Create another list containing only odd numbers.



Sort both lists.



Display both lists.

```python
numbers = [12, 5, 8, 20, 3, 15, 7]

even_numbers = []
odd_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("Even numbers:", sorted(even_numbers))
print("Odd numbers:", sorted(odd_numbers))
```
